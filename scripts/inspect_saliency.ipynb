# Saliency Inspection
Notebook to inspect the generated matrices across all and targeted channels.

```
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

# Add project root to path
sys.path.insert(0, os.path.abspath('..'))

from scripts.asset_scripts.scripts.plot_utils import plot_comparison_figure, plot_first_5_matrices
```

## Global Analysis (All Channels)

```
target = -1
base_dir = "../assets/saliency"

try:
    mat_true = np.load(f"{base_dir}/matrices_true_channel_{target}.npy")
    mat_sal = np.load(f"{base_dir}/saliency_map_channel_{target}.npy")
    mat_ig = np.load(f"{base_dir}/integrated_grads_channel_{target}.npy")

    print(f"Shapes: True {mat_true.shape}, Saliency {mat_sal.shape}, IG {mat_ig.shape}")
except FileNotFoundError:
    print(f"Data for channel {target} not yet generated.")
```

```
# Average over batch and time
mean_true = mat_true.mean(axis=(0, 1))
mean_sal = mat_sal.mean(axis=(0, 1))
mean_ig = mat_ig.mean(axis=(0, 1))

plot_comparison_figure(mean_true, mean_sal, mean_ig, title_prefix=f"Channel {target} Mean")
```

```
# First 5 matrices for sample 0
sample_idx = 0
plot_first_5_matrices(mat_true[sample_idx], f"Channel {target} - True Transfer (Sample {sample_idx})")
plot_first_5_matrices(mat_sal[sample_idx], f"Channel {target} - Saliency Map (Sample {sample_idx})")
plot_first_5_matrices(mat_ig[sample_idx], f"Channel {target} - Integrated Gradients (Sample {sample_idx})")
```

## Targeted Analysis (Channel 0)

```
target = 0

try:
    mat_true = np.load(f"{base_dir}/matrices_true_channel_{target}.npy")
    mat_sal = np.load(f"{base_dir}/saliency_map_channel_{target}.npy")
    mat_ig = np.load(f"{base_dir}/integrated_grads_channel_{target}.npy")
    
    # Average over batch and time
    mean_true = mat_true.mean(axis=(0, 1))
    mean_sal = mat_sal.mean(axis=(0, 1))
    mean_ig = mat_ig.mean(axis=(0, 1))

    plot_comparison_figure(mean_true, mean_sal, mean_ig, title_prefix=f"Channel {target} Mean")
    
    # First 5 matrices for sample 0
    sample_idx = 0
    plot_first_5_matrices(mat_true[sample_idx], f"Channel {target} - True Transfer (Sample {sample_idx})")
    plot_first_5_matrices(mat_sal[sample_idx], f"Channel {target} - Saliency Map (Sample {sample_idx})")
    plot_first_5_matrices(mat_ig[sample_idx], f"Channel {target} - Integrated Gradients (Sample {sample_idx})")
except FileNotFoundError:
    print(f"Data for channel {target} not yet generated.")
```
