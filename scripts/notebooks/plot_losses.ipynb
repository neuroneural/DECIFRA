```python
models_to_plot = [
    # "/Users/ppopov1/DECIFRA/assets/logs/1_pretrain-ukb-GRU_forecaster-default",
    # "/Users/ppopov1/DECIFRA/assets/logs/1_pretrain-ukb-VAR-lag_20_full",
    "/Users/ppopov1/DECIFRA/assets/logs/1_pretrain-ukb-DECIFRA_IMix",
    "/Users/ppopov1/DECIFRA/assets/logs/1_pretrain-ukb-DECIFRA_MS-noMixed",
    "/Users/ppopov1/DECIFRA/assets/logs/1_pretrain-ukb-DECI  FRA_MS-default",
]
# Force a reload of the module
importlib.reload(assets.scripts.plot_losses)
from assets.scripts.plot_losses import plot_model_losses
decifra_df = plot_model_losses(
    models_to_plot, 
    target_loss='val_forecast_loss', 
    ylim=(0.42, 0.50) 
)
decifra_df.headf
```

```python
# 1. Identify the runs to exclude
target_model = '1_pretrain-ukb-DECIFRA_MS-default'

runs_to_exclude = decifra_df[
    (decifra_df['model'] == target_model) & 
    (decifra_df['epoch'] == 499) & 
    (decifra_df['val_forecast_loss'] > 0.45)
]['run'].unique()

print(f"Filtering out runs: {runs_to_exclude}")

# 2. Filter the dataframe to KEEP everything EXCEPT the bad runs for that model
filtered_df = decifra_df[
    ~((decifra_df['model'] == target_model) & (decifra_df['run'].isin(runs_to_exclude)))
]

# Now you can use filtered_df for your plotting
# decifra_df = filtered_df

combined_df = filtered_df
target_loss = "val_forecast_loss"
ylim=(0.42, 0.46)
title=None

last_epoch = combined_df['epoch'].max()
last_stats = combined_df[combined_df['epoch'] == last_epoch].groupby('model')[target_loss].mean().sort_values()
run_counts = combined_df.groupby('model')['run'].nunique()
hue_order = last_stats.index.tolist()

# 2. General Style & Accessibility
sns.set_context("talk") # Global font scale increase
sns.set_palette("colorblind") # Colorblind-friendly palette

fig, ax = plt.subplots(figsize=(16, 9))
# Traces
sns.lineplot(data=combined_df, x='epoch', y=target_loss, palette="colorblind", hue='model', hue_order=hue_order, units='run', estimator=None, alpha=0.4)
# Mean line for each model
sns.lineplot(data=combined_df, x='epoch', y=target_loss, palette="colorblind", hue='model', hue_order=hue_order, legend=False, linewidth=3)

if ylim is not None:
    ax.set_ylim(*ylim)

# 3. Add Table Summary in Center-Top
table_data = [[m, run_counts[m], f"{v:.4f}"] for m, v in last_stats.items()]
# Center-top position: [left, bottom, width, height]
# Placing it at the top center with bbox
stats_table = ax.table(cellText=table_data, 
                        colLabels=['Model', 'Runs', f'Mean {target_loss}'], 
                        loc='center',
                        cellLoc='left',
                        bbox=[0.25, 0.65, 0.5, 0.25])

stats_table.auto_set_font_size(False)
stats_table.set_fontsize(12)
stats_table.set_zorder(100) # Ensure table is in front

# Ensure background is white and opaque
for (row, col), cell in stats_table.get_celld().items():
    cell.set_facecolor('white')
    cell.set_alpha(1.0)
    cell.set_edgecolor('black')
    cell.set_linewidth(1)
    # Adjust column widths: first column much wider
    if col == 0:
        cell.set_width(0.6)
    elif col == 1:
        cell.set_width(0.15)
    else:
        cell.set_width(0.25)

# 4. Optional Legend (Restored as requested)
plt.legend(loc='lower left', prop={'size': 12}, title="Models")

# 5. Labels and Title
plt.title(title if title else f"{target_loss} across Models", fontsize=18, pad=40)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.ylabel(target_loss, fontsize=14)
plt.xlabel('Epoch', fontsize=14)
plt.tight_layout()
plt.show()
```
